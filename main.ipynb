## **Analysis and Data Cleaning**

##### **BATTING**

```python
import pandas as pd
```

```python
# Batting Stats
batting = pd.read_csv("batting_stats.csv")
batting.head()
```

```python
batting.shape
```

```python
batting['out/ not_out'] = batting['dismissal'].apply(lambda x: "not out" if x == 'not out' else "out")
```

```python
batting.drop('dismissal',axis=1, inplace=True)
batting.drop('minutes',axis=1, inplace=True)
```

```python
batting.columns
```

```python
batting['batsman'] = batting['batsman'].str.replace(r'†', '', regex=True)
```

```python
batting.sample(5)
```

```python
print(f"Shape: {batting.shape}")
print("===========================")
print(f"Checking datatypes: \n\n {batting.dtypes}")
```

```python
# Changing dtype of strike_rate
# batting['strike_rate'] = batting['strike_rate'].astype(float)
batting["strike_rate"] = pd.to_numeric(batting["strike_rate"], errors='coerce')
```

```python
batting['strike_rate'] = batting['strike_rate'].fillna(0)
```

```python
batting.dtypes
```

```python
batting.isnull().sum()
```

```python
batting.to_csv('batting_.csv', index=False)
```
