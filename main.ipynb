```python
import time
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup
import pandas as pd

options = Options()
options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--window-size=1920,1080")
options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/124.0.0.0 Safari/537.36")

driver = webdriver.Chrome(options=options)

URLS = [
    ("1",  "Janakpur Bolts vs Kathmandu Gorkhas",     "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/janakpur-bolts-npl-vs-kathmandu-gorkhas-npl-1st-match-1510977/full-scorecard"),
    ("2",  "Chitwan Rhinos vs Karnali Yaks",           "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/chitwan-rhinos-npl-vs-karnali-yaks-npl-2nd-match-1510978/full-scorecard"),
    ("3",  "Biratnagar Kings vs Pokhara Avengers",     "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/biratnagar-kings-npl-vs-pokhara-avengers-npl-3rd-match-1510979/full-scorecard"),
    ("4",  "Kathmandu Gorkhas vs Sudur Paschim Royals","https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/kathmandu-gorkhas-npl-vs-sudur-paschim-royals-npl-4th-match-1510980/full-scorecard"),
    ("5",  "Chitwan Rhinos vs Lumbini Lions",           "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/chitwan-rhinos-npl-vs-lumbini-lions-npl-5th-match-1510981/full-scorecard"),
    ("6",  "Pokhara Avengers vs Sudur Paschim Royals",  "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/pokhara-avengers-npl-vs-sudur-paschim-royals-npl-6th-match-1510982/full-scorecard"),
    ("7",  "Karnali Yaks vs Lumbini Lions",             "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/karnali-yaks-npl-vs-lumbini-lions-npl-7th-match-1510983/full-scorecard"),
    ("8",  "Biratnagar Kings vs Kathmandu Gorkhas",     "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/biratnagar-kings-npl-vs-kathmandu-gorkhas-npl-8th-match-1510984/full-scorecard"),
    ("9",  "Biratnagar Kings vs Janakpur Bolts",        "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/biratnagar-kings-npl-vs-janakpur-bolts-npl-9th-match-1510985/full-scorecard"),
    ("10", "Karnali Yaks vs Sudur Paschim Royals",     "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/karnali-yaks-npl-vs-sudur-paschim-royals-npl-10th-match-1510986/full-scorecard"),
    ("11", "Kathmandu Gorkhas vs Lumbini Lions",        "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/kathmandu-gorkhas-npl-vs-lumbini-lions-npl-11th-match-1510987/full-scorecard"),
    ("12", "Biratnagar Kings vs Chitwan Rhinos",        "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/biratnagar-kings-npl-vs-chitwan-rhinos-npl-12th-match-1510988/full-scorecard"),
    ("13", "Lumbini Lions vs Sudur Paschim Royals",     "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/lumbini-lions-npl-vs-sudur-paschim-royals-npl-13th-match-1510989/full-scorecard"),
    ("14", "Janakpur Bolts vs Pokhara Avengers",        "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/janakpur-bolts-npl-vs-pokhara-avengers-npl-14th-match-1510990/full-scorecard"),
    ("15", "Chitwan Rhinos vs Kathmandu Gorkhas",       "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/chitwan-rhinos-npl-vs-kathmandu-gorkhas-npl-15th-match-1510991/full-scorecard"),
    ("16", "Biratnagar Kings vs Karnali Yaks",          "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/biratnagar-kings-npl-vs-karnali-yaks-npl-16th-match-1510992/full-scorecard"),
    ("17", "Lumbini Lions vs Pokhara Avengers",         "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/lumbini-lions-npl-vs-pokhara-avengers-npl-17th-match-1510993/full-scorecard"),
    ("18", "Janakpur Bolts vs Sudur Paschim Royals",    "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/janakpur-bolts-npl-vs-sudur-paschim-royals-npl-18th-match-1510994/full-scorecard"),
    ("19", "Karnali Yaks vs Kathmandu Gorkhas",         "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/karnali-yaks-npl-vs-kathmandu-gorkhas-npl-19th-match-1510995/full-scorecard"),
    ("20", "Chitwan Rhinos vs Janakpur Bolts",          "https://www.espncricinfo.com/series/nepal-premier-league-2025-26-1510976/chitwan-rhinos-npl-vs-janakpur-bolts-npl-20th-match-1510996/full-scorecard"),
]

all_batting = []
all_bowling = []
all_results = []

for match_no, match_name, url in URLS:
    print(f"Match {match_no}: {match_name}")
    try:
        driver.get(url)
        WebDriverWait(driver, 20).until(EC.presence_of_element_located((By.TAG_NAME, "table")))
        time.sleep(3)

        soup = BeautifulSoup(driver.page_source, "html.parser")
        tables = soup.find_all("table")

        # get result text
        result = ""
        for el in soup.select("p, span, div"):
            t = el.text.strip()
            if "won by" in t.lower() and len(t) < 150:
                result = t
                break

        bat_inn = bowl_inn = 1

        for table in tables:
            first_header = table.select_one("thead th")
            if not first_header:
                continue
            first_header_text = first_header.text.strip().lower()

            # BATTING: first column header is "Batting"
            if first_header_text == "batting":
                rows = table.select("tbody tr")
                for row in rows:
                    cells = [td.text.strip() for td in row.select("td")]
                    # skip extras, total, fall of wickets rows
                    if len(cells) >= 7 and cells[0] not in ("", "Extras", "TOTAL", "Total", "Fall of wickets", "Did not bat"):
                        all_batting.append({
                            "match_no":    match_no,
                            "match":       match_name,
                            "innings":     bat_inn,
                            "batsman":     cells[0],
                            "dismissal":   cells[1],
                            "runs":        cells[2],
                            "balls":       cells[3],
                            "minutes":     cells[4],
                            "4s":          cells[5],
                            "6s":          cells[6],
                            "strike_rate": cells[7] if len(cells) > 7 else "",
                        })
                bat_inn += 1

            # BOWLING: first column header is "Bowling"
            elif first_header_text == "bowling":
                rows = table.select("tbody tr")
                for row in rows:
                    cells = [td.text.strip() for td in row.select("td")]
                    if len(cells) >= 5 and cells[0]:
                        all_bowling.append({
                            "match_no": match_no,
                            "match":    match_name,
                            "innings":  bowl_inn,
                            "bowler":   cells[0],
                            "overs":    cells[1],
                            "maidens":  cells[2],
                            "runs":     cells[3],
                            "wickets":  cells[4],
                            "economy":  cells[5] if len(cells) > 5 else "",
                            "wides":    cells[7] if len(cells) > 7 else "",
                            "noballs":  cells[8] if len(cells) > 8 else "",
                        })
                bowl_inn += 1

        all_results.append({"match_no": match_no, "match": match_name, "result": result, "url": url})
        print(f"  batting={bat_inn-1} innings  bowling={bowl_inn-1} innings")

    except Exception as e:
        print(f"  ERROR: {e}")
        all_results.append({"match_no": match_no, "match": match_name, "result": "Error", "url": url})

driver.quit()

batting_df = pd.DataFrame(all_batting)
bowling_df = pd.DataFrame(all_bowling)
results_df = pd.DataFrame(all_results)
players_df = pd.DataFrame({"player": sorted(
    set(batting_df["batsman"].tolist() if not batting_df.empty else []) |
    set(bowling_df["bowler"].tolist()  if not bowling_df.empty else [])
)})

batting_df.to_csv("batting_stats.csv", index=False)
bowling_df.to_csv("bowling_stats.csv", index=False)
results_df.to_csv("match_results.csv", index=False)
players_df.to_csv("players.csv",       index=False)

print("\n===== BATTING =====")
print(batting_df.head(15))
print("\n===== BOWLING =====")
print(bowling_df.head(15))
print("\n===== RESULTS =====")
print(results_df)
print("\n===== PLAYERS =====")
print(players_df)
print(f"\nDone! batting={len(batting_df)} rows, bowling={len(bowling_df)} rows, players={len(players_df)}")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

players = [
    # ── Original players ──────────────────────────────────
    "Aaditya Mahata","Aakash Chand","Aakash Tripathi","Aarif Sheikh","Aasif Sheikh",
    "Abhisekh Tiwari","Abhishesh Gautam","Abinash Bohara","Adam Rossington","Amar Routela",
    "Anil Sah","Arjun Gharti","Arjun Saud","Basir Ahamad","Ben Charlesworth",
    "Bhim Sharki","Bikash Aagri","Binod Bhandari","Bipin Khatri","Bipin Sharma",
    "Chris Lynn","D'Arcy Short","Dan Douthwaite","Dawid Malan","Deepak Bohara",
    "Deepak Dumre","Dev Khanal","Dilip Nath","Dinesh Kharel","Dipendra Singh Airee",
    "Dipesh Kandel","George Munsey","Gerhard Erasmus","Gulbadin Naib","Gulsan Jha",
    "Harmeet Singh","Hemant Dhami","Ishan Pandey","JJ Smit","James Neesham",
    "Jan Nicol Loftie-Eaton","John Simpson","Josh Brown","Kamal Airee","Karan KC",
    "Kiran Thagunna","Kishor Mahato","Kushal Bhurtel","Kushal Malla","Lahiru Milantha",
    "Lahiru Samarakoon","Lalit Rajbanshi","Lokesh Bam","Maaz Sadaqat","Marchant de Lange",
    "Mark Watt","Martin Guptill","Matt Taylor","Max O'Dowd","Mayan Yadav",
    "Milind Kumar","Mohammad Aadil Alam","Najibullah Zadran","Nandan Yadav","Narayan Joshi",
    "Naren Bhatta","Naren Saud","Niroshan Dickwella","Pawan Sarraf","Pratik Shrestha",
    "Pratis GC","Priyank Panchal","Ranjit Kumar","Rashid Khan","Ravi Bopara",
    "Ricardo Vasconcelos","Rijan Dhakal","Rohit Paudel","Ruben Trumpelmann","Rupesh Singh",
    "Sachin Bhatt","Sagar Dhakal","Sahil Patel","Saif Zaib","Sam Heazlett",
    "Sameer Ali Musalman","Sandeep Lamichhane","Sangeeth Cooray","Sanjay Krishnamurthi","Santosh Yadav",
    "Saugat Dhakal","Scott Kuggeleijn","Shahab Alam","Sher Malla","Shubh Kansakar",
    "Shubham Ranjane","Sohail Tanvir","Sompal Kami","Sumeet Verma","Sumit Maharjan",
    "Sundeep Jora","Sunny Patel","Surya Tamang","Thomas Draca","Tilak Bhandari",
    "Trit Raj Das","Unish Singh","Wayne Parnell","William Bosisto","Yuvraj Khatri",
    "Alpesh Ramjani","Arjun Kumal","Dhananjaya Lakshan","Dhruv Parashar",
    "Dinesh Adhikari","Dipak Bohara","Dipendra Rawat","Faf du Plessis",
    "Gautam KC","George Scrimshaw","Hikmat Mahara","Imran Sheikh",
    "Jason Roy","Jaykishan Kolsawala","Milan Bohara","Movin Subasingha",
    "Mukhtar Ahmed","Puneet Mehra","Rishi Dhawan","Rohit Chand",
    "Sandeep Chettri","Shankar Rana","Sheldon Jackson","Subash Bhandari","Tul Thapa",
]

results = []

for i, name in enumerate(players, 1):
    print(f"[{i}/{len(players)}] {name}")

    info = {
        "player": name,
        "batting_style": "",
        "bowling_style": "",
        "playing_role": "",
        "team": "",
        "country": "",
        "image_url": ""
    }

    wiki_url = f"https://en.wikipedia.org/wiki/{name.replace(' ', '_')}"

    try:
        r = requests.get(wiki_url, headers={'User-Agent': 'Mozilla/5.0'}, timeout=5)
        soup = BeautifulSoup(r.text, 'html.parser')

        infobox = soup.find('table', class_='infobox')

        if infobox:
            img = infobox.find('img')
            if img and img.get('src'):
                img_url = img['src']
                if img_url.startswith('//'):
                    img_url = 'https:' + img_url
                info['image_url'] = img_url

            rows = infobox.find_all('tr')

            for row in rows:
                th = row.find('th')
                td = row.find('td')

                if not th or not td:
                    continue

                label = th.text.strip()
                value = td.get_text(separator=' ').strip()
                value = value.split('[')[0].strip()

                if label == "Batting":
                    info['batting_style'] = value
                elif label == "Bowling":
                    info['bowling_style'] = value
                elif label == "Role":
                    info['playing_role'] = value
                elif "National side" in label or label == "Team":
                    info['team'] = value
                elif label == "Born":
                    lines = value.split('\n')
                    if len(lines) > 1:
                        info['country'] = lines[-1]

            print(f"  ✓ bat={info['batting_style'][:20]} bowl={info['bowling_style'][:20]} img={'YES' if info['image_url'] else 'NO'}")
        else:
            print(f"  ✗ No Wikipedia page")

    except Exception as e:
        print(f"  ✗ Error: {e}")

    results.append(info)
    time.sleep(0.5)

df = pd.DataFrame(results)
df.to_csv("players_full_info.csv", index=False)

print("\n===== SUMMARY =====")
print(f"Total players: {len(df)}")
print(f"With batting info: {df['batting_style'].ne('').sum()}")
print(f"With bowling info: {df['bowling_style'].ne('').sum()}")
print(f"With images: {df['image_url'].ne('').sum()}")
print(f"\nSaved: players_full_info.csv")

print("\n===== SAMPLE =====")
print(df[df['batting_style'] != ''].head(10).to_string())
```
